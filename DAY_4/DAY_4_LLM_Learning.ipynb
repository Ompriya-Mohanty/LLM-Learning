# Day 4 - Retrieval-Augmented Generation (RAG)

## 1. What is RAG?

RAG stands for Retrieval-Augmented Generation.

RAG is a technique where an LLM first retrieves relevant information from an external knowledge source and then uses that information to generate an answer.

Instead of relying only on the knowledge stored in the model, RAG provides relevant external information to the model.

## 2. Why Does RAG Exist?

LLMs have limitations. They may not know private company information, newly created information, or information stored in internal documents.

RAG helps by retrieving relevant information from an external knowledge base and providing it to the LLM.

For example, if a company has 100 internal PDF documents, we can build a RAG system that retrieves relevant sections from those documents when a user asks a question.

Basic flow:

User Question
      ↓
Retrieve Relevant Information
      ↓
Add Retrieved Information to Prompt
      ↓
LLM Generates Answer

## 3. RAG vs Normal LLM

Without RAG:

User → LLM → Answer

The model mainly uses the information available in its context and learned parameters.

With RAG:

User → Retrieval System → Relevant Documents → LLM → Answer

RAG allows the application to provide relevant external information to the model before generating the answer.

## 4. What are Embeddings?

Embeddings are numerical representations of text.

A piece of text is converted into a vector containing numbers.

Texts with similar meanings tend to have vectors that are closer together in the embedding space.

For example:

"How can I reset my password?"

and

"I forgot my password. How do I change it?"

have similar meanings, so their embeddings should be relatively similar.

```python
text1 = "How can I reset my password?"
text2 = "I forgot my password. How do I change it?"

print(text1)
print(text2)
```

## 5. What is Chunking?

Documents can be very large, so we usually do not send the entire document to the LLM.

Chunking means splitting a large document into smaller pieces called chunks.

Example:

Large PDF
   ↓
Document
   ↓
Chunk 1
Chunk 2
Chunk 3
Chunk 4

Smaller chunks make it easier for the retrieval system to find the relevant part of a document.

```python
text = """
RAG is used with large language models.
It retrieves relevant information from external sources.
The retrieved information is provided to the model.
"""

chunk_size = 50

chunks = [
    text[i:i + chunk_size]
    for i in range(0, len(text), chunk_size)
]

for i, chunk in enumerate(chunks):
    print(f"Chunk {i + 1}:")
    print(chunk)
    print()
```

## 6. Chunking Strategies

Different documents may require different chunking strategies.

Common approaches include:

- Fixed-size chunks
- Sentence-based chunks
- Paragraph-based chunks
- Overlapping chunks

Chunk size is important because chunks that are too small may lose context, while chunks that are too large may contain unnecessary information.

The goal is to create chunks that contain useful and meaningful information for retrieval.

## 7. What is a Vector Database?

A vector database stores vector representations, such as embeddings, and allows us to search for similar vectors.

In a RAG system, document chunks are converted into embeddings and stored in a vector database.

When the user asks a question, the question is also converted into an embedding.

The vector database then searches for chunks with similar embeddings.

Document
   ↓
Chunks
   ↓
Embeddings
   ↓
Vector Database

## 8. What is Similarity Search?

Similarity search finds the stored chunks that are most similar to the user's query.

Example:

Query:
"How do I reset my password?"

The retrieval system may find:

"To reset your password, go to Settings → Security → Reset Password."

This chunk is relevant because its meaning is similar to the user's question.

## 9. Cosine Similarity

Cosine similarity is one method used to measure similarity between vectors.

A higher similarity score generally means the vectors are more similar in direction.

This is useful for comparing the embedding of a user query with embeddings of document chunks.

## 10. RAG Pipeline

A basic RAG pipeline has three main stages:

### 1. Retrieval
Find relevant information from the knowledge base.

### 2. Augmentation
Add the retrieved information to the user's prompt or context.

### 3. Generation
Send the question together with the retrieved context to the LLM and generate the final answer.

Overall flow:

Documents
   ↓
Load
   ↓
Chunk
   ↓
Embed
   ↓
Store
   ↓
User Query
   ↓
Retrieve Relevant Chunks
   ↓
Augment Prompt
   ↓
LLM
   ↓
Answer

```python
documents = [
    "Password can be reset from the Security settings.",
    "The billing team handles payment-related issues.",
    "Technical support handles application errors."
]

query = "Where can I reset my password?"

for document in documents:
    if "password" in document.lower():
        print("Relevant document:")
        print(document)
```

## 12. Augmentation

After retrieving relevant information, we add that information to the user's prompt.

For example:

User Question:
"Where can I reset my password?"

Retrieved Context:
"Password can be reset from the Security settings."

The application then creates a prompt containing both the context and the question.

```python
context = "Password can be reset from the Security settings."
question = "Where can I reset my password?"

prompt = f"""
Use the following context to answer the question.

Context:
{context}

Question:
{question}
"""

print(prompt)
```

## 13. Generation

The final step is generation.

The retrieved context and user question are provided to the LLM.

The LLM uses the retrieved context to generate the final answer.

## 15. Common RAG Failure Points

A RAG system can fail at different stages.

### 1. Poor Chunking
If documents are split incorrectly, important context may be lost.

### 2. Poor Retrieval
The system may retrieve irrelevant chunks.

### 3. Incorrect Embeddings
If the embedding representation does not capture the meaning well, similarity search may return poor results.

### 4. Too Much Retrieved Information
If too many irrelevant chunks are added, the LLM may become confused.

### 5. Missing Information
If the required information is not present in the knowledge base, retrieval cannot provide a useful answer.

### 6. Generation Problems
Even with good retrieval, the LLM may misunderstand or incorrectly use the retrieved information.
