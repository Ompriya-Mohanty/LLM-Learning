1. What is Prompt Engineering?

Prompt Engineering is the process of designing and improving
instructions given to an LLM to get useful, accurate and
consistent responses.

A good prompt usually contains:

1. Role
2. Task
3. Context
4. Constraints
5. Output format

```python
prompt = """
You are a Python tutor.

Task:
Explain Python to a beginner.

Context:
The student has no programming experience.

Constraints:
- Use simple English.
- Give 5 bullet points.

Output:
Bullet points only.
"""

print(prompt)
```

2. Prompt Structure

A well-designed prompt can be structured as:

Role + Task + Context + Constraints + Output Format

Role:
Defines how the model should behave.

Task:
Defines what the model needs to do.

Context:
Provides additional information.

Constraints:
Defines rules and limitations.

Output Format:
Defines how the response should be returned.

# 3. Clear and Specific Instructions

LLMs work better when instructions are clear and specific.

A vague prompt may produce a broad or unpredictable response because the model does not know exactly what the user expects.

For example:

"Tell me about Python."

This prompt does not specify the audience, length, style or information required.

A more specific prompt is:

"Explain Python to a complete beginner in 5 simple bullet points. Mention its common uses and provide one example."

Specific instructions help the model understand exactly what needs to be done.

### Key Points

- Clearly define the task.
- Mention the required information.
- Specify the expected length.
- Specify the target audience.
- Specify the desired output format.

```python
vague_prompt = "Tell me about Python."

specific_prompt = """
Explain Python to a complete beginner.

Requirements:
- Use simple English.
- Give exactly 5 bullet points.
- Mention common uses of Python.
- Give one simple example.
"""

print("Vague Prompt:")
print(vague_prompt)

print("\nSpecific Prompt:")
print(specific_prompt)
```

# 4. System, Developer and User Prompts

LLM applications can use different types of instructions to control the behaviour of the model.

System instructions define the general behaviour or purpose of the assistant.

Developer instructions define application-specific rules and requirements.

User instructions contain the actual request made by the user.

### Example

System:
You are a helpful assistant.

Developer:
Always answer in simple English.

User:
Explain Machine Learning.

The model uses these instructions together to generate an appropriate response.

```python
messages = [
    {
        "role": "system",
        "content": "You are a helpful assistant."
    },
    {
        "role": "developer",
        "content": "Always answer in simple English."
    },
    {
        "role": "user",
        "content": "Explain Machine Learning."
    }
]

for message in messages:
    print(message["role"].upper() + ":")
    print(message["content"])
    print()
```

# 5. Delimiters

Delimiters are special characters or markers used to clearly separate instructions from input data.

They help the model understand which part of the prompt contains instructions and which part contains information to process.

Common delimiters include:

- ###
- Triple quotation marks
- XML-style tags such as <text> and </text>

### Example

The instruction can be separated from the customer message using:

### CUSTOMER MESSAGE

My order arrived damaged.

### END CUSTOMER MESSAGE

Using delimiters is especially useful when working with long text, emails, documents or user-provided content.

```python
customer_email = """
Hello,

My order arrived yesterday but the product is damaged.
I would like a replacement.

Thank you.
"""

prompt = f"""
Summarize the customer's problem in one sentence.

### CUSTOMER EMAIL
{customer_email}
### END CUSTOMER EMAIL
"""

print(prompt)
```

# 6. Zero-Shot Prompting

Zero-Shot Prompting means asking an LLM to perform a task without providing any examples.

The model receives an instruction and the input, and it uses its existing knowledge to perform the task.

### Example

Task:

Classify the following sentence as Positive, Negative or Neutral.

Input:

"The product quality is excellent."

The model should return:

Positive

Zero-shot prompting is simple and useful when the task is straightforward and the expected behaviour is easy to understand.

```python
text = "The product quality is excellent and I am very happy."

prompt = f"""
Classify the sentiment as Positive, Negative or Neutral.

Return only the sentiment.

### TEXT
{text}
### END TEXT
"""

print(prompt)
```

# 7. Few-Shot Prompting

Few-Shot Prompting means providing one or more examples in the prompt before asking the model to perform the actual task.

The examples help the model understand the expected pattern, style or format.

### Example

Example 1:
Text: I love this product.
Sentiment: Positive

Example 2:
Text: The product is terrible.
Sentiment: Negative

Example 3:
Text: The package arrived today.
Sentiment: Neutral

Now classify:

Text: The camera quality is amazing.

Expected answer:

Positive

Few-shot prompting is useful when the task is more complex or when a specific output pattern is required.

```python
prompt = """
Classify the sentiment as Positive, Negative or Neutral.

Example 1:
Text: I love this product.
Sentiment: Positive

Example 2:
Text: The product is terrible.
Sentiment: Negative

Example 3:
Text: The package arrived today.
Sentiment: Neutral

Now classify:

Text: The camera quality is amazing.
Sentiment:
"""

print(prompt)
```

# 8. Role Prompting

Role Prompting means assigning a specific role or persona to the LLM.

The role helps define how the model should behave and communicate.

Examples of roles include:

- Python teacher
- Customer support assistant
- Data analyst
- Code reviewer
- Technical interviewer
- Content writer

### Example

"You are a senior Python developer. Review the following Python code and explain the error in simple language."

Role prompting can help make the response more relevant to the task.

```python
prompt = """
You are a senior Python developer.

Review the following code:

numbers = [1, 2, 3]
print(numbers[5])

Explain:
1. What is wrong?
2. Why does the error happen?
3. How can it be fixed?
"""

print(prompt)
```

# 9. Providing Context

Context is additional information provided to the model so that it can understand the task correctly.

Without enough context, the model may generate a generic answer or misunderstand the user's intention.

### Example

Customer message:

"My laptop stopped connecting to Wi-Fi this morning. I have an important meeting at 3 PM."

The context allows the model to understand both the technical problem and its urgency.

### Key Point

Good context should be:

- Relevant
- Accurate
- Sufficient
- Clearly separated from instructions

```python
customer_message = """
My laptop stopped connecting to Wi-Fi this morning.
I have an important meeting at 3 PM.
"""

prompt = f"""
You are a customer support assistant.

Analyze the customer message.

### CUSTOMER MESSAGE
{customer_message}
### END CUSTOMER MESSAGE

Identify:
1. The problem
2. The urgency
3. The requested action
"""

print(prompt)
```

# 10. Constraints

Constraints are rules or limitations that tell the model what it should or should not do.

Examples of constraints include:

- Use simple English.
- Give exactly 5 bullet points.
- Keep the answer below 100 words.
- Return only JSON.
- Do not invent information.
- Use a professional tone.

### Example

"Explain Artificial Intelligence in exactly 5 bullet points using simple English and keep the answer below 100 words."

Constraints make the output more predictable and controlled.

```python
prompt = """
Explain Artificial Intelligence.

Constraints:
- Use simple English.
- Maximum 100 words.
- Give exactly 5 bullet points.
- Do not use complicated technical terms.
"""

print(prompt)
```

# 11. Output Format

Output Format defines how the model should structure its final response.

Common output formats include:

- Bullet points
- Numbered lists
- Tables
- JSON
- Paragraphs
- Emails
- Short answers

### Example

Instead of asking:

"Tell me about the customer."

We can ask:

"Return the customer's name, age and profession in the following format:

Name:
Age:
Profession:"

Specifying the output format makes the response easier to read and process programmatically.

```python
text = """
Rahul is 24 years old and works as a software developer.
"""

prompt = f"""
Extract the following information from the text.

### TEXT
{text}
### END TEXT

Return the answer in this format:

Name:
Age:
Profession:
"""

print(prompt)
```

# 12. Chain-of-Thought Prompting

Chain-of-Thought refers to prompting a model to reason through a problem in a step-by-step manner.

For practical applications, it is often better to request a concise explanation, calculation or reasoning summary rather than asking the model to expose private internal reasoning.

### Example

Question:

A product costs ₹1000 and has a 20% discount. What is the final price?

A useful prompt can ask:

"Calculate the discount and provide the final answer with the calculation."

Expected result:

Discount = ₹200

Final Price = ₹800

This approach is useful for mathematical problems, logical tasks and multi-step problems.

```python
question = """
A shop gives a 20% discount on a product priced at ₹1000.
What is the final price?
"""

prompt = f"""
Solve the problem carefully.

Question:
{question}

Return:
1. The calculation
2. The final answer
"""

print(prompt)
```

# 14. Iterative Prompting

Iterative Prompting means improving a prompt by testing its output and modifying the instructions.

The basic process is:

Write Prompt → Run → Evaluate → Improve → Run Again

A first prompt may produce an incomplete or unclear answer.

After evaluating the result, we can add:

- More context
- Better instructions
- Constraints
- Examples
- Output format

### Example

Initial prompt:

"Summarize this email."

Improved prompt:

"Summarize this email in exactly 3 bullet points. Include the main problem, requested action and urgency. Do not add information that is not present."

Iterative prompting helps improve the reliability and quality of LLM responses.

```python
email = """
My laptop stopped connecting to Wi-Fi this morning.
I have an important meeting at 3 PM and need the issue fixed before then.
"""

prompt_v1 = f"""
Summarize this email:

{email}
"""

prompt_v2 = f"""
Summarize this email in exactly 3 bullet points.

Include:
- Main problem
- Requested action
- Urgency

Do not add information that is not present.

### EMAIL
{email}
### END EMAIL
"""

print("PROMPT 1:")
print(prompt_v1)

print("\nPROMPT 2:")
print(prompt_v2)
```

# 17. Transforming

Transforming means changing existing content into another format, style or language while keeping its original meaning.

Examples include:

- Informal text to professional text
- Paragraph to bullet points
- English to another language
- Text to JSON
- Long text to short text

### Example

Original:

"hey sir i cant come office tomorrow because i have some work"

Transformed:

"Dear Sir, I would like to inform you that I will be unable to come to the office tomorrow due to personal work."

Transformation is useful for communication, content creation and data processing.

```python
text = """
hey sir i cant come office tomorrow because i have some work
"""

prompt = f"""
Rewrite this message as a professional office email.

Keep the original meaning unchanged.

### TEXT
{text}
### END TEXT
"""

print(prompt)
```

# 18. Expanding

Expanding means taking a short idea or piece of information and turning it into a more detailed response.

For example, a short idea such as:

"Write an email asking for one day leave."

Can be expanded into a complete professional email with:

- Subject
- Greeting
- Reason
- Leave request
- Closing

Expansion is useful for content writing, email generation, documentation and other text-generation tasks.

```python
idea = "Write an email asking for one day leave."

prompt = f"""
Expand this idea into a professional email.

Include:
- Subject
- Greeting
- Reason
- Leave request
- Closing

### IDEA
{idea}
### END IDEA
"""

print(prompt)
```

# 19. Prompt Failure Modes

Prompt Failure Modes are common reasons why an LLM may produce a poor or unexpected response.

Common failure modes include:

1. Vague instructions
2. Missing context
3. Conflicting instructions
4. Unclear output format
5. Ambiguous input
6. Too many unnecessary instructions
7. Missing examples for complex tasks
8. Asking the model to invent missing information

### Example

Weak prompt:

"Write something about my project."

This prompt is vague because it does not specify the project details, audience, length, tone or required information.

Improved prompt:

"Write a 100-word professional introduction for a college project called Customer Support Chatbot. Explain its purpose and main features using simple English."

A clear prompt reduces the possibility of unwanted output.

# 21. OpenAI API

An API, or Application Programming Interface, allows a Python program to communicate with an OpenAI model.

Using the OpenAI Python SDK, we can send instructions and user input to a model and receive its response.

The basic workflow is:

Python Program → OpenAI API → Model → Response → Python Program

The API key must be stored securely.

The API key should not be written directly inside the notebook or uploaded to a public GitHub repository.

An environment variable such as OPENAI_API_KEY can be used to store the API key securely.

# 22. Prompt Evaluation

Prompt Evaluation means testing a prompt to determine how well it performs a task.

Important evaluation criteria include:

### Accuracy

Is the response correct?

### Relevance

Does the response answer the actual question?

### Completeness

Does the response contain all required information?

### Consistency

Does the prompt produce reliable results for similar inputs?

### Format Compliance

Does the response follow the requested format?

### Hallucination

Did the model generate information that was not supported by the provided input?

Prompt evaluation helps identify weaknesses and improve prompts.
