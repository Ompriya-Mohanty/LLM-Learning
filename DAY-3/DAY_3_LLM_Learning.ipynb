## 4. Messages Format

LLM applications often work with conversational messages.

Messages can represent different roles, such as:

- System: Defines the behaviour or instructions for the assistant.
- User: Contains the user's request.
- Assistant: Contains the model's response.

This structure allows an application to maintain a conversation and provide instructions to the model.

```python
messages = [
    {
        "role": "system",
        "content": "You are a helpful assistant."
    },
    {
        "role": "user",
        "content": "What is an LLM?"
    }
]

print(messages)
```

## 6. Structured Outputs

LLMs normally generate natural-language responses.

For example, instead of returning only a sentence, we can represent the information as:

## 8. Streaming

Normally, an API request returns the response after the model has generated the response.

With streaming, the application can receive parts of the response while the model is generating it.

Without streaming:

Request → Wait → Complete Response

With streaming:

Request → Part 1 → Part 2 → Part 3 → Complete Response

Streaming is useful for long responses because the user can start seeing the response before the entire response has been generated.

The exact streaming implementation depends on the API and SDK being used.

## 9. Token Limits

LLMs process text using tokens.

A token can represent a complete word, part of a word, punctuation, or other pieces of text.

An API request can contain input tokens and generate output tokens.

The model has a context limit, which determines how much information can be processed within a request.

Basic idea:

Input Tokens + Output Tokens = Total Tokens Used

If the input becomes too large, it may exceed the model's context limit.

Therefore, developers need to manage the amount of text sent to the model.

```python
text = "Large Language Models process text using tokens."

words = text.split()

print("Text:", text)
print("Approximate word count:", len(words))
```

## 10. API Cost

Using an LLM API can have a cost depending on the provider and model.

The cost is commonly related to the number of input and output tokens used.

Conceptually:

Cost = Input Token Cost + Output Token Cost

Using unnecessarily large prompts or generating unnecessarily long responses can increase token usage and therefore increase cost.

Actual pricing depends on the selected model and provider, so current pricing should always be checked from the provider's official documentation.

```python
input_tokens = 500
output_tokens = 200

total_tokens = input_tokens + output_tokens

print("Input tokens:", input_tokens)
print("Output tokens:", output_tokens)
print("Total tokens:", total_tokens)
```
